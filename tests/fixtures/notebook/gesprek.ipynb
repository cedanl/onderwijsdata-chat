## Vraag 1

Hoeveel eerstejaars hbo-studenten kozen per studiejaar voor techniek?

### Antwoord

Per studiejaar kozen tussen 2021 en 2025 elk jaar enkele duizenden eerstejaars hbo-studenten voor techniek.

**Bronnen**
- DUO · Eerstejaars ingeschrevenen hbo (p02ho1ejrs)

> Let op: de aantallen zijn een ondergrens: DUO onderdrukt kleine cellen

#### Herkomst

Verantwoording: model Testmodel · seed niet toegepast · prompt abc123def456 · catalogus bd4942907ddd · app 2.8.0 (commit 7d3488aa)

Dataversie: DUO p02ho1ejrs (herladen 26-11-2026)

Stap 1, "Catalogus doorzocht", heeft geen code: voor die stap maakt de app geen code om zelf na te rekenen.

```python
# Stap 2: Data gefilterd
import pandas as pd
from onderwijsbronnen import duo

df = duo.load("p02ho1ejrs", 0)
# DUO-sentinel -1 = onderdrukte cel: DUO publiceert kleine aantallen niet, geen telwaarde. De app sluit ze uit.
# Totalen over een selectie met zulke cellen zijn een ondergrens.
for kolom in df.select_dtypes('number').columns:
    if not any(h in str(kolom).upper() for h in ('MUTATIE', 'SALDO', 'VERSCHIL', 'GROEI', 'DELTA')):
        df[kolom] = df[kolom].mask(df[kolom].isin([-1]))
if "STUDIEJAAR" in df.columns:
    df["STUDIEJAAR_LABEL"] = df["STUDIEJAAR"].map(lambda j: f"{j}/{j + 1}")
df = df[df['ONDERDEEL'].astype(str).str.lower().isin(['techniek'])]
group_by = ['STUDIEJAAR']
if "STUDIEJAAR_LABEL" in df.columns and "STUDIEJAAR_LABEL" not in group_by:
    group_by.append("STUDIEJAAR_LABEL")
for kolom in ['AANTAL_EERSTEJAARS_INGESCHREVENEN']:
    df[kolom] = pd.to_numeric(df[kolom], errors="coerce")
df = df.groupby(group_by, dropna=False).agg({'AANTAL_EERSTEJAARS_INGESCHREVENEN': lambda s: s.sum(min_count=1)}).reset_index()
print(df)
```

## Vraag 2

Reken het zelf na met een eigen analyse.

### Antwoord

Een eigen analyse op hetzelfde bestand geeft dezelfde aantallen.

**Bronnen**
- DUO · Eerstejaars ingeschrevenen hbo (p02ho1ejrs)

#### Herkomst

Verantwoording: model Testmodel · seed niet toegepast · prompt abc123def456 · catalogus bd4942907ddd · app 2.8.0 (commit 7d3488aa)

Dataversie: DUO p02ho1ejrs (herladen 26-11-2026)

```python
# Stap 1: Analyse uitgevoerd
import pandas as pd
from onderwijsbronnen import duo

df = duo.load("p02ho1ejrs", 0)
# DUO-sentinel -1 = onderdrukte cel: DUO publiceert kleine aantallen niet, geen telwaarde. De app sluit ze uit.
# Totalen over een selectie met zulke cellen zijn een ondergrens.
for kolom in df.select_dtypes('number').columns:
    if not any(h in str(kolom).upper() for h in ('MUTATIE', 'SALDO', 'VERSCHIL', 'GROEI', 'DELTA')):
        df[kolom] = df[kolom].mask(df[kolom].isin([-1]))
if "STUDIEJAAR" in df.columns:
    df["STUDIEJAAR_LABEL"] = df["STUDIEJAAR"].map(lambda j: f"{j}/{j + 1}")
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

techniek = df[df["ONDERDEEL"].astype(str).str.lower() == "techniek"]
result = techniek.groupby("STUDIEJAAR", as_index=False)["AANTAL_EERSTEJAARS_INGESCHREVENEN"].sum()
print(result)
```
